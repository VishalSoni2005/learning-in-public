```python

# Assignment 2: CBOW Model using NumPy

import numpy as np

# Training sentence
sentence = "i love deep learning and artificial intelligence"
words = sentence.split()

# Create vocabulary
vocab = sorted(set(words))
word_to_id = {w: i for i, w in enumerate(vocab)}
vocab_size = len(vocab)

# Generate context-target pairs
data = []
for i in range(2, len(words) - 2):
    context = [words[i-2], words[i-1], words[i+1], words[i+2]]
    target = words[i]
    data.append((context, target))

# Initialize weights
np.random.seed(42)
embedding = np.random.randn(vocab_size, 10) * 0.1
output_weights = np.random.randn(10, vocab_size) * 0.1

# Train CBOW model
for epoch in range(200):
    for context, target in data:

        # Convert context words to indices
        ids = [word_to_id[w] for w in context]
        target_id = word_to_id[target]

        # Average context embeddings
        hidden = np.mean(embedding[ids], axis=0)

        # Softmax prediction
        scores = np.dot(hidden, output_weights)
        exp_scores = np.exp(scores - np.max(scores))
        probs = exp_scores / np.sum(exp_scores)

        # Calculate error
        error = probs.copy()
        error[target_id] -= 1

        # Update output weights
        output_weights -= 0.05 * np.outer(hidden, error)

        # Update embeddings
        grad = np.dot(output_weights, error)
        for idx in ids:
            embedding[idx] -= 0.05 * grad / len(ids)

# Test prediction
context = ["i", "love", "and", "artificial"]
ids = [word_to_id[w] for w in context]

hidden = np.mean(embedding[ids], axis=0)
scores = np.dot(hidden, output_weights)
prediction = np.argmax(scores)

print("Context Words:", context)
print("Predicted Target Word:", vocab[prediction])
print("Vocabulary:", vocab)
```
