```python

# Assignment 1: TensorFlow and PyTorch

import tensorflow as tf
import torch
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# ---------------- TENSORFLOW ----------------
print("===== TENSORFLOW =====")

# Create tensors
a = tf.constant([1, 2, 3, 4]) # -> tensors are multidimensional arrays, similar to NumPy arrays
b = tf.constant([5, 6, 7, 8])

# Tensor operations
print("Tensor A:", a.numpy())
print("Tensor B:", b.numpy())
print("Addition:", (a + b).numpy())
print("Multiplication:", (a * b).numpy())

# Simple neural network
model = Sequential([
    Dense(8, activation='relu', input_shape=(4,)),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy')

model.fit([[1, 2, 3, 4], [4, 3, 2, 1]],
          [[1], [0]], epochs=5, verbose=0)

print("TensorFlow Prediction:",
      model.predict([[1, 2, 3, 4]], verbose=0))


# ---------------- PYTORCH ----------------
print("\n===== PYTORCH =====")

# Create tensors
x = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
y = torch.tensor([5, 6, 7, 8], dtype=torch.float32)

# Tensor operations
print("Tensor X:", x)
print("Tensor Y:", y)
print("Addition:", x + y)
print("Multiplication:", x * y)

# Simple neural network
net = torch.nn.Sequential(
    torch.nn.Linear(4, 8),
    torch.nn.ReLU(),
    torch.nn.Linear(8, 1),
    torch.nn.Sigmoid()
)

# Prediction
output = net(torch.tensor([[1., 2., 3., 4.]]))

print("PyTorch Prediction:", output.detach().numpy())
```
