```python
import numpy as np

# Create actual values: 45 ones, 52 zeros, 8 ones = total 105 elements
actual = list(np.ones(45)) + list(np.zeros(52)) + list(np.ones(8))
actual = np.array(actual)

# Create predicted values: 40 ones, 52 zeros, 8 ones = total 100 elements
predicted = list(np.ones(40)) + list(np.zeros(52)) + list(np.ones(8))
predicted = np.array(predicted)
```

```python
print("Actual length:", len(actual))
print("Predicted length:", len(predicted))
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

print("Accuracy:", accuracy_score(actual, predicted))
print("Confusion Matrix:\n", confusion_matrix(actual, predicted))
```

```python
import numpy as np
from sklearn.metrics import accuracy_score, confusion_matrix

# Define 'actual' labels
actual = list(np.ones(45)) + list(np.zeros(52)) + list(np.ones(8))  # total = 105

# Define 'predicted' labels (same length)
predicted = list(np.ones(40)) + list(np.zeros(52)) + list(np.ones(13))  # also total = 105

# Convert to numpy arrays
actual = np.array(actual)
predicted = np.array(predicted)

# Evaluate
print("Accuracy:", accuracy_score(actual, predicted))
print("Confusion Matrix:\n", confusion_matrix(actual, predicted))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score

actual = list(np.ones(45)) + list(np.zeros(52)) + list(np.ones(8))      # Total = 105
predicted = list(np.ones(40)) + list(np.zeros(52)) + list(np.ones(13)) # Total = 105

actual = np.array(actual)
predicted = np.array(predicted)

cm = confusion_matrix(actual, predicted)


plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Pred 0', 'Pred 1'], yticklabels=['Actual 0', 'Actual 1'])
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[0, 1])
disp.plot(cmap='Blues')
plt.show()
```
