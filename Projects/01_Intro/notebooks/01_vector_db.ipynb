```python
data = collection.get(
    include=["documents", "embeddings", "metadatas"]
)

for i, emb in enumerate(data["embeddings"]):
    doc_id = data["ids"][i]
    metadata = data["metadatas"][i]
    doc_text = data["documents"][i].strip()

    print(f"ID: {doc_id}")
    print(f"Metadata: {metadata}")
    print(f"Embedding Dimension: {len(emb)}")
    print(f"Embedding Preview (first 5 dimensions): {emb[:5]}")
    print(f"Document Snippet: {doc_text[:70]}...")
    print("Vishal")
    print("-" * 50)
```
