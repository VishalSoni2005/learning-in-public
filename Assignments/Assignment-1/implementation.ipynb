```python
# Deep Learning Practical 1: TensorFlow and Keras

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 1. TensorFlow: Tensor operations
a, b = tf.constant(15), tf.constant(20)
print("TensorFlow addition:", (a + b).numpy())

# 2. TensorFlow: Train a simple linear model
x = np.random.rand(100).astype(np.float32)
y = 0.2 * x + 0.2

W = tf.Variable(tf.random.normal([1]))
bias = tf.Variable(tf.zeros([1]))
optimizer = tf.keras.optimizers.Adam(learning_rate=0.05)

for step in range(1000):
    with tf.GradientTape() as tape:
        y_pred = W * x + bias
        loss = tf.reduce_mean(tf.square(y_pred - y))
    gradients = tape.gradient(loss, [W, bias])
    optimizer.apply_gradients(zip(gradients, [W, bias]))

print("TensorFlow learned weight:", W.numpy())
print("TensorFlow learned bias:", bias.numpy())
print("Final MSE:", loss.numpy())

# 3. Keras: Binary classification using synthetic data
np.random.seed(42)
x = np.random.rand(200, 8).astype(np.float32)
y = (x[:, 0] + x[:, 1] > 1).astype(np.float32)

model = Sequential([
    tf.keras.Input(shape=(8,)),
    Dense(12, activation="relu"),
    Dense(8, activation="relu"),
    Dense(1, activation="sigmoid")
])

model.compile(optimizer="adam", loss="binary_crossentropy",
              metrics=["accuracy"])

model.fit(x, y, epochs=10, batch_size=10, verbose=0)
loss, accuracy = model.evaluate(x, y, verbose=0)
print("Keras loss:", round(loss, 4))
print("Keras accuracy:", round(accuracy, 4))
print("Sample prediction:", model.predict(x[:1], verbose=0)[0][0])
```
